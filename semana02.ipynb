```python
import pandas as pd

df = pd.read_excel('dados_semana02.xlsx')

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar o estilo dos gráficos
sns.set_style('whitegrid')
```

### 1. Contagem de Amostras por Sabor

Vamos começar visualizando a distribuição dos diferentes sabores presentes nos dados.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Sabor', data=df, palette='viridis')
plt.title('Contagem de Amostras por Sabor')
plt.xlabel('Sabor')
plt.ylabel('Número de Amostras')
plt.show()
```

### 2. Distribuição de °Brix por Sabor

Agora, vamos analisar como o °Brix (grau de doçura) varia entre os diferentes sabores.

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='Sabor', y='°Brix', data=df, palette='plasma')
plt.title('Distribuição de °Brix por Sabor')
plt.xlabel('Sabor')
plt.ylabel('°Brix')
plt.show()
```

### 3. Distribuição de pH por Sabor

Podemos também verificar a distribuição do pH para cada sabor.

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='Sabor', y='pH', data=df, palette='cividis')
plt.title('Distribuição de pH por Sabor')
plt.xlabel('Sabor')
plt.ylabel('pH')
plt.show()
```

### 4. °Brix Médio por Sabor (Gráfico de Barras)

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Sabor', y='°Brix', data=df, estimator=lambda x: x.mean(), palette='viridis')
plt.title('°Brix Médio por Sabor')
plt.xlabel('Sabor')
plt.ylabel('°Brix Médio')
plt.show()
```

### 5. Distribuição Geral de °Brix (Box Plot)

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y='°Brix', data=df, palette='plasma')
plt.title('Distribuição Geral de °Brix')
plt.ylabel('°Brix')
plt.show()
```

### 6. Distribuição de °Brix (Histograma)

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['°Brix'], bins=5, kde=True, color='skyblue')
plt.title('Histograma de Distribuição de °Brix')
plt.xlabel('°Brix')
plt.ylabel('Frequência')
plt.show()
```

### 7. Distribuição de pH (Histograma)

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['pH'], bins=5, kde=True, color='lightcoral')
plt.title('Histograma de Distribuição de pH')
plt.xlabel('pH')
plt.ylabel('Frequência')
plt.show()
```

### 8. Distribuição de Sabores (Gráfico de Pizza)

```python
plt.figure(figsize=(9, 9))
df['Sabor'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title('Distribuição Percentual de Sabores')
plt.ylabel('') # Remove o rótulo 'Sabor' do eixo Y que é adicionado automaticamente pelo pandas
plt.show()
```
